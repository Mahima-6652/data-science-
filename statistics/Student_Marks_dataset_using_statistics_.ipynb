```python
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn as sb

print("library imported")
```

```python
df = pd.read_csv('Student_Marks.csv')
df.head(5)
```

```python
df.tail(5)
```

```python
# select marks col
marks = df.select_dtypes(include="number")
```

```python
# mean calculate
marks.mean()
```

```python
# median calculate
marks.median()
```

```python
# Calculate mode
marks.mode().iloc[0]
```

```python
# Calculate variance
marks.var()
```

```python
# Calculate standard deviation
marks.std()
```

```python
# Calculate percentiles
marks.quantile([0.25, 0.50, 0.75, 0.90])
```

```python
# Calculate total marks for each student
df["Total_Marks"] = marks.sum(axis=1)
```

```python
# Calculate average marks for each student
df["Average_Marks"] = marks.mean(axis=1)
```

```python
# the top 10 students
top_students = df.nlargest(10, "Total_Marks")
```

```python
print("TOP 10 STUDENTS")
display(top_students[
    ["student_name", "Total_Marks", "Average_Marks"]
])
```

```python
#the bottom 10 students
bottom_students = df.nsmallest(10, "Total_Marks")
```

```python
display(bottom_students[
    ["student_name", "Total_Marks", "Average_Marks"]
])
```

```python
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)

IQR = Q3 - Q1

print("INTERQUARTILE RANGE")
print(IQR)
```

```python
print("SKEWNESS")
marks.skew()
```

```python
print("KURTOSIS")
marks.kurt()
```

```python
five_number_summary = pd.DataFrame({
    "Minimum": marks.min(),
    "Q1": marks.quantile(0.25),
    "Median": marks.median(),
    "Q3": marks.quantile(0.75),
    "Maximum": marks.max()
})

display(five_number_summary)
```

```python
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = (marks < lower_limit) | (marks > upper_limit)

print("Potential outliers in each subject:")
print(outliers.sum())
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))

sb.boxplot(data=marks)

plt.title("Distribution of Marks Across Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=30)
plt.show()
```

```python
statistics_summary = marks.describe().T

statistics_summary["variance"] = marks.var()
statistics_summary["skewness"] = marks.skew()
statistics_summary["kurtosis"] = marks.kurt()
statistics_summary["range"] = marks.max() - marks.min()
statistics_summary["IQR"] = (
    marks.quantile(0.75) - marks.quantile(0.25)
)

display(statistics_summary.round(2))
```

```python
# Calculate total and average marks
df["Total_Marks"] = marks.sum(axis=1)
df["Average_Marks"] = marks.mean(axis=1)

# Overall class statistics
print("Class average marks:", round(df["Average_Marks"].mean(), 2))
print("Highest student average:", round(df["Average_Marks"].max(), 2))
print("Lowest student average:", round(df["Average_Marks"].min(), 2))
print("Median student average:", round(df["Average_Marks"].median(), 2))
```

```python
subject_averages = marks.mean().sort_values(ascending=False)

print("Subject-wise average marks:")
display(subject_averages)
```
